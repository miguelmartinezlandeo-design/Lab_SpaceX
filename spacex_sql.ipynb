##### Enumera los registros que mostrarán los nombres de los meses, los resultados de fallos de aterrizaje en el barco dron, las versiones del propulsor y el sitio de lanzamiento para los meses del año 2015.

```python
# La columna 'Date' contiene la fecha; no existen columnas 'YEAR' ni 'MONTH' en df
df["Date"] = pd.to_datetime(df["Date"])

result = df.loc[
	(df["Date"].dt.year == 2015) &
	(df["Landing_Outcome"] == "Failure (drone ship)"),
	["Date", "Landing_Outcome", "Booster_Version", "Launch_Site"]
].copy()

result["MONTH"] = result["Date"].dt.month_name()
result = result[["MONTH", "Date", "Landing_Outcome", "Booster_Version", "Launch_Site"]]

result
```

#### sql
SELECT DATE_FORMAT(Date, '%M') AS MONTH, Date, Landing_Outcome, Booster_Version, Launch_Site FROM SPACEXTBL WHERE YEAR(Date) = '2015' AND Landing_Outcome = 'Failure (drone ship)';<br>
or <br>
SELECT substr(Date, 6, 2) AS MONTH, Date, Landing_Outcome, Booster_Version, Launch_Site FROM SPACEXTBL WHERE substr(Date, 1, 4) = '2015' AND Landing_Outcome = 'Failure (drone ship)';

### Tarea 10

##### Clasifica el recuento de los resultados de aterrizaje (como Fallo (barco dron) o Éxito (plataforma terrestre)) entre las fechas 2010-06-04 y 2017-03-20, en orden descendente.

```python
##### Clasifica el recuento de los resultados de aterrizaje (como Fallo (barco dron) o Éxito (plataforma terrestre)) entre las fechas 2010-06-04 y 2017-03-20, en orden descendente.
df["Date"] = pd.to_datetime(df["Date"])
result = df.loc[
    (df["Date"] >= "2010-06-04") &
    (df["Date"] <= "2017-03-20"),
    ["Landing_Outcome"]
].copy()

result = result.groupby("Landing_Outcome").size().reset_index(name="count").sort_values(by="count", ascending=False)

result
# sql
```

#### sql 
SELECT Landing_Outcome, COUNT(*) AS count FROM SPACEXTBL WHERE Date >= '2010-06-04' AND Date <= '2017-03-20' GROUP BY Landing_Outcome ORDER BY count DESC;
